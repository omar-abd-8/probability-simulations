```python
%matplotlib widget
import numpy as np
import matplotlib.pyplot as plt
from collections import defaultdict
from ipywidgets import (FloatSlider, Dropdown, ToggleButtons, Checkbox,
                        VBox, HBox, Layout, Label, HTML, Output)

# ── Simulation parameters ──
FS_SIM = 500                                    # kHz – for smooth curves
T      = 1.4                                    # ms – display window
N_SIM  = int(FS_SIM * T)
t      = np.linspace(0, T, N_SIM, endpoint=False)
N_REP  = 5                                      # spectral replicas: −N … +N
F_MAX  = 35                                     # kHz – frequency axis limit

# ── Widgets: source signal ──
signal_dd = Dropdown(options=['Cosine', 'Sine'], value='Cosine',
                     description='Signal:', layout=Layout(width='180px'))
amp_sl    = FloatSlider(value=0.7, min=0.1, max=2.0, step=0.1,
                        description='A:', readout_format='.2f',
                        style={'description_width': '35px'},
                        layout=Layout(width='280px'))
freq_sl   = FloatSlider(value=4.5, min=0.5, max=12.0, step=0.5,
                        description='f₀ [kHz]:', readout_format='.1f',
                        style={'description_width': '70px'},
                        layout=Layout(width='280px'))
phase_sl  = FloatSlider(value=45, min=0, max=180, step=5,
                        description='φ [°]:', readout_format='.0f',
                        style={'description_width': '35px'},
                        layout=Layout(width='280px'))

# ── Widgets: sampling & reconstruction ──
fa_sl = FloatSlider(value=10.0, min=1.0, max=30.0, step=0.5,
                    description='fₐ [kHz]:', readout_format='.1f',
                    style={'description_width': '70px'},
                    layout=Layout(width='280px'))
fe_sl = FloatSlider(value=5.0, min=0.5, max=15.0, step=0.5,
                    description='fₑ [kHz]:', readout_format='.1f',
                    style={'description_width': '70px'},
                    layout=Layout(width='280px'))

# ── Widgets: display ──
spectrum_mode = ToggleButtons(options=['Real / Imaginary', 'Magnitude'],
                              value='Magnitude', description='',
                              style={'button_width': '90px'})
cb_xt  = Checkbox(value=True,  description='x(t)',  indent=False, layout=Layout(width='90px'))
cb_xa  = Checkbox(value=False,  description='xₐ(t)', indent=False, layout=Layout(width='90px'))
cb_xr  = Checkbox(value=False,  description='x̃(t)',  indent=False, layout=Layout(width='90px'))
cb_Xf  = Checkbox(value=True,  description='X(f)',  indent=False, layout=Layout(width='90px'))
cb_XAf = Checkbox(value=False,  description='Xₐ(f)', indent=False, layout=Layout(width='90px'))
cb_Hf  = Checkbox(value=False,  description='Hₑ(f)', indent=False, layout=Layout(width='90px'))

info_html = HTML(value='')

# ── Helpers ──

def spectral_lines(sig_type, A, f0, phi):
    """Analytical Fourier transform → (frequencies, complex coefficients)."""
    if sig_type == 'Cosine':
        return (np.array([-f0, f0]),
                np.array([(A/2)*np.exp(-1j*phi), (A/2)*np.exp(1j*phi)]))
    else:
        return (np.array([-f0, f0]),
                np.array([(1j*A/2)*np.exp(-1j*phi), (-1j*A/2)*np.exp(1j*phi)]))


def replicate(freqs, coeffs, fa):
    """Create spectral replicas at multiples of fₐ."""
    ff, cc = [], []
    for n in range(-N_REP, N_REP + 1):
        ff.append(freqs + n * fa)
        cc.append(coeffs.copy())
    return np.concatenate(ff), np.concatenate(cc)


def combine_lines(freqs, coeffs):
    """Sum complex coefficients at the same frequency."""
    bucket = defaultdict(complex)
    for fk, ck in zip(freqs, coeffs):
        bucket[round(fk, 6)] += ck
    f_out = np.array(list(bucket.keys()))
    c_out = np.array(list(bucket.values()))
    order = np.argsort(f_out)
    return f_out[order], c_out[order]


def extract(coeffs, mode):
    """Select real / imaginary / magnitude."""
    if mode == 'Magnitude':  return np.abs(coeffs)
    return np.real(coeffs), np.imag(coeffs)


def style_axes(ax):
    """Hide top/right spines, keep bottom/left thin, add bold x/y axis lines."""
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)
    ax.spines['bottom'].set_linewidth(0.8)
    ax.spines['left'].set_linewidth(0.8)
    ax.axhline(0, color='black', linewidth=1.2, zorder=0)
    ax.axvline(0, color='black', linewidth=1.2, zorder=0)


# ── Figure ──
out1 = Output()
with out1:
    fig, (ax_t, ax_f) = plt.subplots(2, 1, figsize=(9, 6),
                                      gridspec_kw={'hspace': 0.50})
    fig.canvas.header_visible = False
    plt.show()


def update(_=None):
    A     = amp_sl.value
    f0    = freq_sl.value
    phi   = np.deg2rad(phase_sl.value)
    fa    = fa_sl.value
    fe    = fe_sl.value
    stype = signal_dd.value
    mode  = spectrum_mode.value

    func = np.cos if stype == 'Cosine' else np.sin

    # ── Time-domain signals ──
    x = A * func(2*np.pi*f0*t + phi)

    t_s = np.arange(0, T, 1.0/fa)
    x_s = A * func(2*np.pi*f0*t_s + phi)

    # ── Spectra ──
    base_f, base_c = spectral_lines(stype, A, f0, phi)
    raw_f, raw_c   = replicate(base_f, base_c, fa)

    # Combine overlapping spectral lines (handles aliasing correctly)
    samp_f, samp_c = combine_lines(raw_f, raw_c)

    # ── Reconstruction: lowpass at ±fₑ ──
    mask  = np.abs(samp_f) <= fe + 1e-9
    rec_f = samp_f[mask]
    rec_c = samp_c[mask]

    # Reconstructed signal: x̃(t) = Σ Re[cₖ e^{i2πfₖt}]
    x_rec = np.zeros_like(t)
    for fk, ck in zip(rec_f, rec_c):
        x_rec += np.real(ck * np.exp(1j*2*np.pi*fk*t))

    # ── Powers ──
    Px  = np.mean(x**2)
    Pv  = np.mean((x_rec - x)**2)
    snr = 10*np.log10(Px / Pv) if Pv > 1e-12 else np.inf

    # ── Time-domain plot ──
    ax_t.cla()
    if cb_xt.value:
        ax_t.plot(t, x, color='steelblue', lw=1.5, label='x(t)')
    if cb_xa.value and len(t_s):
        ml, sl, bl = ax_t.stem(t_s, x_s, linefmt='C2-', markerfmt='C2o',
                               basefmt='none')
        sl.set_linewidth(1.2); ml.set_markersize(4)
        ax_t.plot([], [], 'C2o-', label='xₐ(t)')
    if cb_xr.value:
        ax_t.plot(t, x_rec, 'r--', lw=1.3, label='x̃(t)')

    ax_t.set_xlim(0, T);  ax_t.set_ylim(-2.5, 2.5)
    ax_t.set_xlabel('t [ms]')
    ax_t.set_ylabel('Amplitude')
    ax_t.set_title('Time Domain')
    ax_t.legend(loc='upper right', fontsize=8)
    ax_t.grid(True, alpha=0.3)
    style_axes(ax_t)

    # ── Frequency-domain plot ──
    ax_f.cla()

    # Sampled spectrum (with overlapping lines already combined)
    if cb_XAf.value and len(samp_f):
        if mode == 'Magnitude':
            vals = extract(samp_c, mode)
            ml, sl, bl = ax_f.stem(samp_f, vals, linefmt='C2-', markerfmt='C2o',
                                   basefmt='none')
            sl.set_linewidth(1.2); ml.set_markersize(5)
            ax_f.plot([], [], 'C2o-', label='Xₐ(f)')
        else:
            re, im = extract(samp_c, mode)
            ml, sl, bl = ax_f.stem(samp_f, re, linefmt='C0-', markerfmt='C0o',
                                   basefmt='none')
            sl.set_linewidth(1.2); ml.set_markersize(5)
            ml2, sl2, bl2 = ax_f.stem(samp_f, im, linefmt='C3-', markerfmt='C3s',
                                      basefmt='none')
            sl2.set_linewidth(1.2); ml2.set_markersize(5)
            ax_f.plot([], [], 'C0o-', label='Re Xₐ(f)')
            ax_f.plot([], [], 'C3s-', label='Im Xₐ(f)')

    # Original spectrum
    if cb_Xf.value:
        if mode == 'Magnitude':
            vals = extract(base_c, mode)
            ml, sl, bl = ax_f.stem(base_f, vals, linefmt='steelblue',
                                   markerfmt='o', basefmt='none')
            sl.set_linewidth(2.5); ml.set_color('steelblue'); ml.set_markersize(8)
            ax_f.plot([], [], 'o-', color='steelblue', label='X(f)')
        else:
            re, im = extract(base_c, mode)
            ml, sl, bl = ax_f.stem(base_f, re, linefmt='C0-', markerfmt='C0o',
                                   basefmt='none')
            sl.set_linewidth(2.5); ml.set_markersize(8)
            ml2, sl2, bl2 = ax_f.stem(base_f, im, linefmt='C3-', markerfmt='C3s',
                                      basefmt='none')
            sl2.set_linewidth(2.5); ml2.set_markersize(8)
            ax_f.plot([], [], 'C0o-', label='Re X(f)')
            ax_f.plot([], [], 'C3s-', label='Im X(f)')

    # Reconstruction filter
    if cb_Hf.value:
        ax_f.axvspan(-fe, fe, alpha=0.08, color='orange')
        ax_f.axvline(-fe, color='orange', ls='--', lw=1, label='Hₑ(f)')
        ax_f.axvline( fe, color='orange', ls='--', lw=1)

    ax_f.set_xlim(-F_MAX, F_MAX)
    ymax = max(1.2, A/2 + 0.3)
    ax_f.set_ylim((-ymax, ymax) if mode != 'Magnitude' else (-0.05, ymax))
    ax_f.set_xlabel('f [kHz]')
    ax_f.set_ylabel('Magnitude' if mode == 'Magnitude' else 'Value')
    ax_f.set_title(f'Frequency Domain – {mode}')
    ax_f.legend(loc='upper right', fontsize=8)
    ax_f.grid(True, alpha=0.3)
    style_axes(ax_f)

    # ── Info panel ──
    snr_s = f'{snr:.1f} dB' if np.isfinite(snr) else '∞'
    info_html.value = (
        f'<div style="font-size:13px; margin-top:6px">'
        f'<b>Pₓ</b> = {Px:.4f} &emsp; '
        f'<b>Pᵥ</b> = {Pv:.6f} &emsp; '
        f'<b>10·lg(Pₓ/Pᵥ)</b> = {snr_s}</div>'
    )

    fig.canvas.draw_idle()


# ── Connect all widgets ──
for w in [amp_sl, freq_sl, phase_sl, fa_sl, fe_sl, signal_dd,
          spectrum_mode, cb_xt, cb_xa, cb_xr, cb_Xf, cb_XAf, cb_Hf]:
    w.observe(update, names='value')

update()

# ── Layout ──
col_sig  = VBox([Label('Source Signal'), signal_dd, amp_sl, freq_sl, phase_sl])
col_samp = VBox([Label('Sampling / Reconstruction'), fa_sl, fe_sl, info_html])
col_disp = VBox([Label('Display'), spectrum_mode,
                 Label('Time domain:'),  HBox([cb_xt, cb_xa, cb_xr]),
                 Label('Freq. domain:'), HBox([cb_Xf, cb_XAf, cb_Hf])])
controls = VBox([col_sig, col_samp, col_disp],
                layout=Layout(min_width='260px'))

HBox([out1, controls])
```

```python
%matplotlib widget
import numpy as np
import matplotlib.pyplot as plt
from collections import defaultdict
from ipywidgets import (FloatSlider, Dropdown, ToggleButtons, Checkbox,
                        VBox, HBox, Layout, Label, HTML, Output)

# ── Simulation parameters ──
FS_SIM2 = 500
T2      = 1.4
N_SIM2  = int(FS_SIM2 * T2)
t2      = np.linspace(0, T2, N_SIM2, endpoint=False)
N_REP2  = 5
F_MAX2  = 35

# ── Fixed low-frequency carrier ──
A_LO   = 1.0
F_LO   = 2.0    # kHz
PHI_LO = 0.0    # rad
C_LO   = 'darkorange'   # colour for carrier in both domains

# ── Widgets: adjustable high-frequency component ──
amp2_sl   = FloatSlider(value=0.3, min=0.1, max=1.0, step=0.05,
                        description='A_hi:', readout_format='.2f',
                        style={'description_width': '50px'},
                        layout=Layout(width='280px'))
freq2_sl  = FloatSlider(value=8.0, min=1.0, max=12.0, step=0.5,
                        description='f_hi [kHz]:', readout_format='.1f',
                        style={'description_width': '80px'},
                        layout=Layout(width='280px'))
phase2_sl = FloatSlider(value=0, min=0, max=355, step=5,
                        description='φ_hi [°]:', readout_format='.0f',
                        style={'description_width': '50px'},
                        layout=Layout(width='280px'))

# ── Widgets: sampling & reconstruction ──
fa2_sl = FloatSlider(value=20.0, min=1.0, max=40.0, step=0.5,
                     description='fₐ [kHz]:', readout_format='.1f',
                     style={'description_width': '70px'},
                     layout=Layout(width='280px'))
fe2_sl = FloatSlider(value=10.0, min=0.5, max=20.0, step=0.5,
                     description='fₑ [kHz]:', readout_format='.1f',
                     style={'description_width': '70px'},
                     layout=Layout(width='280px'))

# ── Widgets: display ──
spec_mode2 = ToggleButtons(options=['Real / Imaginary', 'Magnitude'],
                           value='Magnitude', description='',
                           style={'button_width': '90px'})
cb2_xt  = Checkbox(value=True,  description='x(t)',  indent=False, layout=Layout(width='90px'))
cb2_xa  = Checkbox(value=False, description='xₐ(t)', indent=False, layout=Layout(width='90px'))
cb2_xr  = Checkbox(value=False, description='x̃(t)',  indent=False, layout=Layout(width='90px'))
cb2_Xf  = Checkbox(value=True,  description='X(f)',  indent=False, layout=Layout(width='90px'))
cb2_XAf = Checkbox(value=False, description='Xₐ(f)', indent=False, layout=Layout(width='90px'))
cb2_Hf  = Checkbox(value=False, description='Hₑ(f)', indent=False, layout=Layout(width='90px'))

info2_html = HTML(value='')

# ── Reuse helpers from cell above ──
# (spectral_lines, replicate, combine_lines, extract, style_axes)

# ── Figure ──
out2 = Output()
with out2:
    fig2, (ax2_t, ax2_f) = plt.subplots(2, 1, figsize=(9, 6),
                                         gridspec_kw={'hspace': 0.50})
    fig2.canvas.header_visible = False
    plt.show()


def update2(_=None):
    A_hi   = amp2_sl.value
    f_hi   = freq2_sl.value
    phi_hi = np.deg2rad(phase2_sl.value)
    fa     = fa2_sl.value
    fe     = fe2_sl.value
    mode   = spec_mode2.value

    # ── Time-domain signals ──
    x_lo = A_LO * np.cos(2*np.pi*F_LO*t2 + PHI_LO)
    x_hi = A_hi * np.cos(2*np.pi*f_hi*t2 + phi_hi)
    x = x_lo + x_hi

    t_s = np.arange(0, T2, 1.0/fa)
    x_s = (A_LO * np.cos(2*np.pi*F_LO*t_s + PHI_LO) +
           A_hi * np.cos(2*np.pi*f_hi*t_s + phi_hi))

    # ── Spectra (separate for colour-coded plotting) ──
    lo_f, lo_c = spectral_lines('Cosine', A_LO, F_LO, PHI_LO)
    hi_f, hi_c = spectral_lines('Cosine', A_hi, f_hi, phi_hi)

    # Combined for sampling / reconstruction
    base_f, base_c = combine_lines(np.concatenate([lo_f, hi_f]),
                                   np.concatenate([lo_c, hi_c]))
    raw_f, raw_c   = replicate(base_f, base_c, fa)
    samp_f, samp_c = combine_lines(raw_f, raw_c)

    # ── Reconstruction: lowpass at ±fₑ ──
    mask  = np.abs(samp_f) <= fe + 1e-9
    rec_f = samp_f[mask]
    rec_c = samp_c[mask]

    x_rec = np.zeros_like(t2)
    for fk, ck in zip(rec_f, rec_c):
        x_rec += np.real(ck * np.exp(1j*2*np.pi*fk*t2))

    # ── Powers ──
    Px  = np.mean(x**2)
    Pv  = np.mean((x_rec - x)**2)
    snr = 10*np.log10(Px / Pv) if Pv > 1e-12 else np.inf

    # ── Time-domain plot ──
    ax2_t.cla()
    if cb2_xt.value:
        ax2_t.plot(t2, x_lo, color=C_LO, ls='--', lw=1.3, alpha=0.7,
                   label=f'carrier ({A_LO}·cos 2π·{F_LO}kHz·t)')
        ax2_t.plot(t2, x, color='steelblue', lw=1.5, label='x(t) = carrier + hi')
    if cb2_xa.value and len(t_s):
        ml, sl, bl = ax2_t.stem(t_s, x_s, linefmt='C2-', markerfmt='C2o',
                                basefmt='none')
        sl.set_linewidth(1.2); ml.set_markersize(4)
        ax2_t.plot([], [], 'C2o-', label='xₐ(t)')
    if cb2_xr.value:
        ax2_t.plot(t2, x_rec, 'r--', lw=1.3, label='x̃(t)')

    ax2_t.set_xlim(0, T2);  ax2_t.set_ylim(-3.0, 3.0)
    ax2_t.set_xlabel('t [ms]')
    ax2_t.set_ylabel('Amplitude')
    ax2_t.set_title('Time Domain')
    ax2_t.legend(loc='upper right', fontsize=8)
    ax2_t.grid(True, alpha=0.3)
    style_axes(ax2_t)

    # ── Frequency-domain plot ──
    ax2_f.cla()

    if cb2_XAf.value and len(samp_f):
        if mode == 'Magnitude':
            vals = extract(samp_c, mode)
            ml, sl, bl = ax2_f.stem(samp_f, vals, linefmt='C2-', markerfmt='C2o',
                                    basefmt='none')
            sl.set_linewidth(1.2); ml.set_markersize(5)
            ax2_f.plot([], [], 'C2o-', label='Xₐ(f)')
        else:
            re, im = extract(samp_c, mode)
            ml, sl, bl = ax2_f.stem(samp_f, re, linefmt='C0-', markerfmt='C0o',
                                    basefmt='none')
            sl.set_linewidth(1.2); ml.set_markersize(5)
            ml2, sl2, bl2 = ax2_f.stem(samp_f, im, linefmt='C3-', markerfmt='C3s',
                                       basefmt='none')
            sl2.set_linewidth(1.2); ml2.set_markersize(5)
            ax2_f.plot([], [], 'C0o-', label='Re Xₐ(f)')
            ax2_f.plot([], [], 'C3s-', label='Im Xₐ(f)')

    if cb2_Xf.value:
        if mode == 'Magnitude':
            # Carrier spectrum in carrier colour
            vals_lo = extract(lo_c, mode)
            ml, sl, bl = ax2_f.stem(lo_f, vals_lo, linefmt=C_LO,
                                    markerfmt='o', basefmt='none')
            sl.set_linewidth(2.5); ml.set_color(C_LO); ml.set_markersize(8)
            ax2_f.plot([], [], 'o-', color=C_LO, label='X_lo(f)')

            vals_hi = extract(hi_c, mode)
            ml, sl, bl = ax2_f.stem(hi_f, vals_hi, linefmt='steelblue',
                                    markerfmt='o', basefmt='none')
            sl.set_linewidth(2.5); ml.set_color('steelblue'); ml.set_markersize(8)
            ax2_f.plot([], [], 'o-', color='steelblue', label='X_hi(f)')
        else:
            # Combined spectrum real/imag
            all_c = np.concatenate([lo_c, hi_c])
            all_f = np.concatenate([lo_f, hi_f])
            re, im = extract(all_c, mode)
            ml, sl, bl = ax2_f.stem(all_f, re, linefmt='C0-', markerfmt='C0o',
                                    basefmt='none')
            sl.set_linewidth(2.5); ml.set_markersize(8)
            ml2, sl2, bl2 = ax2_f.stem(all_f, im, linefmt='C3-', markerfmt='C3s',
                                       basefmt='none')
            sl2.set_linewidth(2.5); ml2.set_markersize(8)
            ax2_f.plot([], [], 'C0o-', label='Re X(f)')
            ax2_f.plot([], [], 'C3s-', label='Im X(f)')

    if cb2_Hf.value:
        ax2_f.axvspan(-fe, fe, alpha=0.08, color='orange')
        ax2_f.axvline(-fe, color='orange', ls='--', lw=1, label='Hₑ(f)')
        ax2_f.axvline( fe, color='orange', ls='--', lw=1)

    ax2_f.set_xlim(-F_MAX2, F_MAX2)
    ymax = max(1.2, max(A_LO, A_hi)/2 + 0.3)
    ax2_f.set_ylim((-ymax, ymax) if mode != 'Magnitude' else (-0.05, ymax))
    ax2_f.set_xlabel('f [kHz]')
    ax2_f.set_ylabel('Magnitude' if mode == 'Magnitude' else 'Value')
    ax2_f.set_title(f'Frequency Domain – {mode}')
    ax2_f.legend(loc='upper right', fontsize=8)
    ax2_f.grid(True, alpha=0.3)
    style_axes(ax2_f)

    # ── Info panel ──
    snr_s = f'{snr:.1f} dB' if np.isfinite(snr) else '∞'
    info2_html.value = (
        f'<div style="font-size:13px; margin-top:6px">'
        f'<b>Pₓ</b> = {Px:.4f} &emsp; '
        f'<b>Pᵥ</b> = {Pv:.6f} &emsp; '
        f'<b>10·lg(Pₓ/Pᵥ)</b> = {snr_s}</div>'
    )

    fig2.canvas.draw_idle()


# ── Connect all widgets ──
for w in [amp2_sl, freq2_sl, phase2_sl, fa2_sl, fe2_sl, spec_mode2,
          cb2_xt, cb2_xa, cb2_xr, cb2_Xf, cb2_XAf, cb2_Hf]:
    w.observe(update2, names='value')

update2()

# ── Layout ──
col2_sig  = VBox([Label('High-Freq Component'), amp2_sl, freq2_sl, phase2_sl,
                  Label(f'Low-Freq Carrier (fixed): A={A_LO}, f={F_LO} kHz')])
col2_samp = VBox([Label('Sampling / Reconstruction'), fa2_sl, fe2_sl, info2_html])
col2_disp = VBox([Label('Display'), spec_mode2,
                  Label('Time domain:'),  HBox([cb2_xt, cb2_xa, cb2_xr]),
                  Label('Freq. domain:'), HBox([cb2_Xf, cb2_XAf, cb2_Hf])])
controls2 = VBox([col2_sig, col2_samp, col2_disp],
                 layout=Layout(min_width='260px'))

HBox([out2, controls2])
```

```python
%matplotlib widget
import numpy as np
import matplotlib.pyplot as plt

# ── Parameters ──
fs   = 20.0     # kHz – sampling rate for simulation
dur  = 10.0     # ms  – signal duration
N    = int(fs * dur)
t4   = np.linspace(0, dur, N, endpoint=False)
f_bw = 4.0      # kHz – bandwidth (hard cutoff)

# ── Generate band-limited signal in frequency domain ──
rng = np.random.default_rng(42)
X_half = np.fft.rfft(rng.standard_normal(N))
f_half = np.fft.rfftfreq(N, d=1.0/fs)       # kHz (one-sided)

# Speech-like spectral shaping: 1/f rolloff + hard band-limit
envelope = np.ones_like(f_half)
envelope[1:] = 1.0 / (1.0 + (f_half[1:] / 0.8)**0.8)
envelope[f_half > f_bw] = 0.0
X_half *= envelope

# Back to time domain, normalise
x4 = np.fft.irfft(X_half, n=N)
x4 = x4 / np.max(np.abs(x4)) * 0.9

# Two-sided spectrum: mirror the one-sided magnitude
mag_half = np.abs(X_half) / N
# Build symmetric two-sided arrays: [-f_max … 0 … +f_max]
f_full = np.concatenate([-f_half[:0:-1], f_half])
mag    = np.concatenate([ mag_half[:0:-1], mag_half])

# ── Plot ──
fig4, (ax4_t, ax4_f) = plt.subplots(2, 1, figsize=(9, 5),
                                     gridspec_kw={'hspace': 0.50})
fig4.canvas.header_visible = False

# Time domain
ax4_t.plot(t4, x4, color='steelblue', lw=0.8)
ax4_t.set_xlim(0, dur)
ax4_t.set_ylim(-1.2, 1.2)
ax4_t.set_xlabel('t [ms]')
ax4_t.set_ylabel('Amplitude')
ax4_t.set_title('Time Domain — Band-Limited Signal (BW = 4 kHz)')
ax4_t.grid(True, alpha=0.3)
style_axes(ax4_t)

# Magnitude spectrum (two-sided)
ax4_f.fill_between(f_full, mag, color='steelblue', alpha=0.3)
ax4_f.plot(f_full, mag, color='steelblue', lw=0.8)
ax4_f.axvline(-f_bw, color='crimson', ls='--', lw=1.2, label=f'±{f_bw} kHz')
ax4_f.axvline( f_bw, color='crimson', ls='--', lw=1.2)
ax4_f.set_xlim(-fs/2, fs/2)
ax4_f.set_ylim(0, None)
ax4_f.set_xlabel('f [kHz]')
ax4_f.set_ylabel('|X(f)|')
ax4_f.set_title('Frequency Domain — Magnitude Spectrum')
ax4_f.legend(loc='upper right', fontsize=8)
ax4_f.grid(True, alpha=0.3)
style_axes(ax4_f)

fig4.canvas.draw_idle()
```

```python
%matplotlib widget
import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import (FloatSlider, ToggleButtons, Checkbox,
                        VBox, HBox, Layout, Label, HTML, Output)

def style_axes(ax):
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)
    ax.spines['bottom'].set_linewidth(0.8)
    ax.spines['left'].set_linewidth(0.8)
    ax.axhline(0, color='black', linewidth=1.2, zorder=0)
    ax.axvline(0, color='black', linewidth=1.2, zorder=0)

# ── Parameters ──
FS5    = 500       # kHz – simulation rate
T5     = 1.4       # ms – display window
N5     = int(FS5 * T5)
t5     = np.linspace(-T5/2, T5/2, N5, endpoint=False)
F_MAX5 = 30        # kHz – frequency display range
N_REP5 = 4         # spectral replicas for sampling
NF     = 2000      # frequency grid points for smooth display

# Dense frequency grid for analytical spectrum
f_d = np.linspace(-F_MAX5, F_MAX5, NF)
df  = f_d[1] - f_d[0]

# ── Fixed axis limits ──
# Time domain: max amplitude is 2.0 (max A slider)
T_YLIM = (-0.5, 2.5)
# Frequency domain: max peak is A_max * W_max = 2.0 * 0.6 = 1.2
F_YLIM_MAG = (-0.05, 1.3)
F_YLIM_SYM = (-1.3, 1.3)

# ── Widgets ──
amp5_sl = FloatSlider(value=1.0, min=0.1, max=2.0, step=0.1,
                      description='A:', readout_format='.1f',
                      style={'description_width': '35px'},
                      layout=Layout(width='280px'))
pos5_sl = FloatSlider(value=0.0, min=-0.5, max=0.5, step=0.01,
                      description='t\u2080 [ms]:', readout_format='.2f',
                      style={'description_width': '70px'},
                      layout=Layout(width='280px'))
wid5_sl = FloatSlider(value=0.4, min=0.02, max=0.6, step=0.02,
                      description='W [ms]:', readout_format='.2f',
                      style={'description_width': '70px'},
                      layout=Layout(width='280px'))

fa5_sl = FloatSlider(value=20.0, min=2.0, max=40.0, step=0.5,
                     description='f\u2090 [kHz]:', readout_format='.1f',
                     style={'description_width': '70px'},
                     layout=Layout(width='280px'))
fe5_sl = FloatSlider(value=10.0, min=0.5, max=20.0, step=0.5,
                     description='f\u2091 [kHz]:', readout_format='.1f',
                     style={'description_width': '70px'},
                     layout=Layout(width='280px'))

spec_mode5 = ToggleButtons(options=['Real / Imaginary', 'Magnitude'],
                           value='Magnitude', description='',
                           style={'button_width': '90px'})

cb5_xt  = Checkbox(value=True,  description='x(t)',  indent=False, layout=Layout(width='90px'))
cb5_xa  = Checkbox(value=False, description='x\u2090(t)', indent=False, layout=Layout(width='90px'))
cb5_xr  = Checkbox(value=False, description='x\u0303(t)',  indent=False, layout=Layout(width='90px'))
cb5_Xf  = Checkbox(value=True,  description='X(f)',  indent=False, layout=Layout(width='90px'))
cb5_XAf = Checkbox(value=False, description='X\u2090(f)', indent=False, layout=Layout(width='90px'))
cb5_Hf  = Checkbox(value=False, description='H\u2091(f)', indent=False, layout=Layout(width='90px'))

info5_html = HTML(value='')

# ── Analytical Fourier transform of rect pulse ──
def rect_ft(f, A, W, t0):
    return A * W * np.sinc(f * W) * np.exp(-1j * 2 * np.pi * f * t0)

# ── Figure ──
out5 = Output()
with out5:
    fig5, (ax5_t, ax5_f) = plt.subplots(2, 1, figsize=(9, 6),
                                         gridspec_kw={'hspace': 0.50})
    fig5.canvas.header_visible = False
    plt.show()


def update5(_=None):
    A   = amp5_sl.value
    t0  = pos5_sl.value
    W   = wid5_sl.value
    fa  = fa5_sl.value
    fe  = fe5_sl.value
    mode = spec_mode5.value

    # ── Time-domain rectangle pulse ──
    x = A * ((t5 >= t0 - W/2) & (t5 < t0 + W/2)).astype(float)

    # Sampled time points
    t_s = np.arange(-T5/2, T5/2, 1.0/fa)
    x_s = A * ((t_s >= t0 - W/2) & (t_s < t0 + W/2)).astype(float)

    # ── Analytical spectra ──
    X_orig = rect_ft(f_d, A, W, t0)

    X_A = np.zeros(NF, dtype=complex)
    for n in range(-N_REP5, N_REP5 + 1):
        X_A += rect_ft(f_d - n * fa, A, W, t0)

    # ── Reconstruction ──
    H = (np.abs(f_d) <= fe).astype(float)
    X_rec = X_A * H

    # Numerical inverse FT for reconstruction
    phase = np.exp(1j * 2 * np.pi * np.outer(t5, f_d))
    x_rec = np.real(phase @ X_rec) * df

    # ── Display helper ──
    def extr(c):
        if mode == 'Magnitude': return np.abs(c)
        return np.real(c), np.imag(c)

    # ── Powers ──
    Px  = np.mean(x**2)
    Pv  = np.mean((x_rec - x)**2)
    snr = 10*np.log10(Px / Pv) if Pv > 1e-12 else np.inf

    # ── Time-domain plot ──
    ax5_t.cla()
    if cb5_xt.value:
        ax5_t.plot(t5, x, color='steelblue', lw=1.5, label='x(t)')
    if cb5_xa.value and len(t_s):
        ml, sl, bl = ax5_t.stem(t_s, x_s, linefmt='C2-', markerfmt='C2o',
                                basefmt='none')
        sl.set_linewidth(1.2); ml.set_markersize(4)
        ax5_t.plot([], [], 'C2o-', label='x\u2090(t)')
    if cb5_xr.value:
        ax5_t.plot(t5, x_rec, 'r--', lw=1.3, label='x\u0303(t)')

    ax5_t.set_xlim(-T5/2, T5/2)
    ax5_t.set_ylim(*T_YLIM)
    ax5_t.set_xlabel('t [ms]')
    ax5_t.set_ylabel('Amplitude')
    ax5_t.set_title('Time Domain')
    ax5_t.legend(loc='upper right', fontsize=8)
    ax5_t.grid(True, alpha=0.3)
    style_axes(ax5_t)

    # ── Frequency-domain plot ──
    ax5_f.cla()

    if cb5_XAf.value:
        if mode == 'Magnitude':
            ax5_f.plot(f_d, extr(X_A), color='C2', lw=1.0, alpha=0.7, label='X\u2090(f)')
        else:
            re, im = extr(X_A)
            ax5_f.plot(f_d, re, color='C0', lw=1.0, alpha=0.7, label='Re X\u2090(f)')
            ax5_f.plot(f_d, im, color='C3', lw=1.0, alpha=0.7, label='Im X\u2090(f)')

    if cb5_Xf.value:
        if mode == 'Magnitude':
            ax5_f.plot(f_d, extr(X_orig), color='steelblue', lw=1.5, label='X(f)')
        else:
            re, im = extr(X_orig)
            ax5_f.plot(f_d, re, color='C0', lw=1.5, label='Re X(f)')
            ax5_f.plot(f_d, im, color='C3', lw=1.5, label='Im X(f)')

    if cb5_Hf.value:
        ax5_f.axvspan(-fe, fe, alpha=0.08, color='orange')
        ax5_f.axvline(-fe, color='orange', ls='--', lw=1, label='H\u2091(f)')
        ax5_f.axvline( fe, color='orange', ls='--', lw=1)

    ax5_f.set_xlim(-F_MAX5, F_MAX5)
    if mode == 'Magnitude':
        ax5_f.set_ylim(*F_YLIM_MAG)
    else:
        ax5_f.set_ylim(*F_YLIM_SYM)
    ax5_f.set_xlabel('f [kHz]')
    ax5_f.set_ylabel('Magnitude' if mode == 'Magnitude' else 'Value')
    ax5_f.set_title('Frequency Domain \u2013 ' + mode)
    ax5_f.legend(loc='upper right', fontsize=8)
    ax5_f.grid(True, alpha=0.3)
    style_axes(ax5_f)

    # ── Info panel ──
    snr_s = f'{snr:.1f} dB' if np.isfinite(snr) else '\u221e'
    info5_html.value = (
        f'<div style="font-size:13px; margin-top:6px">'
        f'<b>P\u2093</b> = {Px:.4f} &emsp; '
        f'<b>P\u1d65</b> = {Pv:.6f} &emsp; '
        f'<b>10\u00b7lg(P\u2093/P\u1d65)</b> = {snr_s}</div>'
    )

    fig5.canvas.draw_idle()


# ── Connect ──
for w in [amp5_sl, pos5_sl, wid5_sl, fa5_sl, fe5_sl, spec_mode5,
          cb5_xt, cb5_xa, cb5_xr, cb5_Xf, cb5_XAf, cb5_Hf]:
    w.observe(update5, names='value')

update5()

# ── Layout ──
col5_sig  = VBox([Label('Rectangle Pulse'), amp5_sl, pos5_sl, wid5_sl])
col5_samp = VBox([Label('Sampling / Reconstruction'), fa5_sl, fe5_sl, info5_html])
col5_disp = VBox([Label('Display'), spec_mode5,
                  Label('Time domain:'),  HBox([cb5_xt, cb5_xa, cb5_xr]),
                  Label('Freq. domain:'), HBox([cb5_Xf, cb5_XAf, cb5_Hf])])
controls5 = VBox([col5_sig, col5_samp, col5_disp],
                 layout=Layout(min_width='260px'))

HBox([out5, controls5])
```

```python
%matplotlib widget
import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import IntSlider, FloatSlider, VBox, HBox, Layout, Label, Output

def style_axes(ax):
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)
    ax.spines['bottom'].set_linewidth(0.8)
    ax.spines['left'].set_linewidth(0.8)
    ax.axhline(0, color='black', linewidth=1.2, zorder=0)
    ax.axvline(0, color='black', linewidth=1.2, zorder=0)

# ── Parameters ──
T6     = 3.0       # periods to show
FS6    = 2000      # samples per period
t6     = np.linspace(-T6/2, T6/2, int(T6 * FS6), endpoint=False)

# ── Widgets ──
n_sl = IntSlider(value=1, min=1, max=80, step=1,
                 description='N (terms):',
                 style={'description_width': '80px'},
                 layout=Layout(width='300px'))
amp_sl6 = FloatSlider(value=1.0, min=0.1, max=2.0, step=0.1,
                      description='A:',
                      readout_format='.1f',
                      style={'description_width': '80px'},
                      layout=Layout(width='300px'))
freq_sl6 = FloatSlider(value=1.0, min=0.5, max=5.0, step=0.5,
                       description='f\u2080 [Hz]:',
                       readout_format='.1f',
                       style={'description_width': '80px'},
                       layout=Layout(width='300px'))

# ── Figure ──
out6 = Output()
with out6:
    fig6, (ax6_t, ax6_f) = plt.subplots(2, 1, figsize=(9, 6),
                                         gridspec_kw={'hspace': 0.50})
    fig6.canvas.header_visible = False
    plt.show()


def update6(_=None):
    N  = n_sl.value
    A  = amp_sl6.value
    f0 = freq_sl6.value

    # ── Ideal square wave ──
    sq = A * np.sign(np.sin(2 * np.pi * f0 * t6))

    # ── Fourier series partial sum ──
    approx = np.zeros_like(t6)
    harmonics = []
    coeffs = []
    for k in range(N):
        n_harm = 2 * k + 1
        a_k = 4 * A / (np.pi * n_harm)
        approx += a_k * np.sin(2 * np.pi * n_harm * f0 * t6)
        harmonics.append(n_harm * f0)
        coeffs.append(a_k)

    # ── Time-domain plot ──
    ax6_t.cla()
    ax6_t.plot(t6, sq, color='gray', lw=0.8, label='Square wave')
    ax6_t.plot(t6, approx, color='steelblue', lw=1.5,
               label=f'Fourier series (N={N})')
    ax6_t.set_xlim(-T6/2, T6/2)
    ax6_t.set_ylim(-2.5, 2.5)
    ax6_t.set_xlabel('t [s]')
    ax6_t.set_ylabel('Amplitude')
    ax6_t.set_title('Time Domain')
    ax6_t.legend(loc='upper right', fontsize=8)
    ax6_t.grid(True, alpha=0.3)
    style_axes(ax6_t)

    # ── Frequency-domain plot (magnitude of Fourier coefficients) ──
    ax6_f.cla()
    harm_arr = np.array(harmonics)
    coeff_arr = np.array(coeffs)

    # Two-sided: mirror to negative frequencies
    f_plot = np.concatenate([-harm_arr[::-1], harm_arr])
    c_plot = np.concatenate([coeff_arr[::-1] / 2, coeff_arr / 2])

    ml, sl, bl = ax6_f.stem(f_plot, c_plot, linefmt='steelblue',
                            markerfmt='o', basefmt='none')
    sl.set_linewidth(2.0)
    ml.set_color('steelblue')
    ml.set_markersize(5)

    # Envelope: 4A/(pi * 2pi*f) = 2A/(pi^2 * f/f0) ... just show 1/f curve
    f_env = np.linspace(f0, (2*N-1)*f0 + f0, 500)
    env = 2 * A / (np.pi * f_env / f0)
    ax6_f.plot(f_env, env, 'r--', lw=0.8, alpha=0.5, label=r'$2A/(\pi n)$ envelope')
    ax6_f.plot(-f_env, env, 'r--', lw=0.8, alpha=0.5)

    f_max = max((2*N-1)*f0 + f0, 5*f0)
    ax6_f.set_xlim(-f_max, f_max)
    ax6_f.set_ylim(-0.05, 2.5)
    ax6_f.set_xlabel('f [Hz]')
    ax6_f.set_ylabel('Magnitude')
    ax6_f.set_title('Fourier Series Coefficients')
    ax6_f.legend(loc='upper right', fontsize=8)
    ax6_f.grid(True, alpha=0.3)
    style_axes(ax6_f)

    fig6.canvas.draw_idle()


# ── Connect ──
for w in [n_sl, amp_sl6, freq_sl6]:
    w.observe(update6, names='value')

update6()

# ── Layout ──
controls6 = VBox([Label('Fourier Series Approximation'),
                  n_sl, amp_sl6, freq_sl6],
                 layout=Layout(min_width='260px'))

HBox([out6, controls6])
```
